```python
def generate_fibonacci(limit):

    sequence = []
    a, b = 0, 1
    while a <= limit:
        sequence.append(a)
        a, b = b, a + b
    return sequence


limit = 100
fib_sequence = generate_fibonacci(limit)
print(f"Fibonacci sequence up to {limit}:")
print(fib_sequence)
```

```python
def rotate_list(lst)
    if not lst:
        return lst

    n = len(lst)
    k = k % n
    return lst[-k:] + lst[:-k]


my_list = [1, 2, 3, 4, 5]

rotated_right = rotate_list(my_list, 2)
print(f"Original list:  {my_list}")
print(f"Rotated right by 2: {rotated_right}")


rotated_left = rotate_list(my_list, -2)
print(f"Rotated left by 2:  {rotated_left}")
```
